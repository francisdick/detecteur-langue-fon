crée une liste d’au moins 8 phrases mélangeant les 3 langues, puis affiche uniquement celles qui contiennent plus de cinq mots. Indice : len(p.split())

```python
# Exercice 2
phrases_trilingues = [
    "Koudo, my friend, comment ça va ce matin ?",
    "Je suis too tired, je vais aller mlɔ́ maintenant.",
    "A fɔ́n ganjí ? I hope you slept well hier soir.",
    "Mi yì grab some food, j’ai trop faim !",
    "Please, avant demain.",
    "Le meeting commence à dix heures, ma dɔ̀ gbɛ̌ ó.",
    "C’est so expensive, un six pour acheter ça aujourd’hui.",
    "Your new outfit est magnifique."
]

for phrase in phrases_trilingues:
    if len(phrase.split()) > 10:
        print(phrase)
```

écris une fonction compter_lettres_fon(phrase) qui renvoie combien de fois ɔ et ɛ apparaissent dans une phrase. Indice : phrase.count("ɔ")

```python
# Exercice 3
def compter_lettres_fon(phrase):
    a = phrase.count("ɔ")
    b = phrase.count("ɛ")
    c = a + b
    print(c)
```
